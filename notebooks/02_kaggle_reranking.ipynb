# Kaggle full-test Reader reranking
Unexecuted recipe. Select GPU T4 x2 and Internet. Choose one dataset per run. Saved completed results and logs are under results/.

```python
from pathlib import Path
import subprocess,sys
PROJECT=Path('/kaggle/working/dpr-50k-reader-reranking')
if not PROJECT.exists():
    subprocess.check_call(['git','clone','https://github.com/ThieuEmAnhKhongSongDuoc/dpr-50k-reader-reranking.git',str(PROJECT)])
subprocess.check_call([sys.executable,'-m','pip','install','-r',str(PROJECT/'requirements-rerank.txt')])
import torch
print(torch.__version__,torch.cuda.device_count())
assert torch.cuda.device_count()>=2,'Select GPU T4 x2'
```

```python
DATA=Path('/kaggle/working/dpr-artifacts')
subprocess.check_call([sys.executable,str(PROJECT/'scripts/download_artifacts.py'),'--group','main','--output',str(DATA)])
```

```python
DATASET='nq_test'  # or curatedtrec_test, trivia_test
CANDIDATES=200
OUTPUT=Path('/kaggle/working')/f'{DATASET}_reader{CANDIDATES}.json'
cmd=[sys.executable,str(PROJECT/'scripts/run_dpr_reader_reranking.py'),
'--reader-checkpoint',str(DATA/'reader.cp'),'--retriever-checkpoint',str(DATA/'retriever.cp'),
'--passage-vectors',str(DATA/'passage_vectors.pkl'),'--corpus',str(DATA/'wiki50k.tsv'),
'--qa-dir',str(DATA/'qa'),'--datasets',DATASET,'--reader-gpus','2',
'--reader-batch-size','16','--query-batch-size','8','--top-docs',str(CANDIDATES),'--output',str(OUTPUT)]
subprocess.run(cmd,cwd=PROJECT,check=True)
print(OUTPUT.read_text())
```
